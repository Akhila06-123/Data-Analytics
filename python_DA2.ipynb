```python
#concat and merging
#joins
import pandas as pd

df1 = pd.DataFrame({
    "EmpID": [101, 102, 103, 104],
    "Name": ["Akhila", "Ravi", "Priya", "Arun"],
    "DeptID": [1, 2, 3, 4]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "DeptID": [2, 3, 4, 5],
    "Department": ["IT", "HR", "Sales", "Finance"]
})

print(df2)
```

```python
inner = pd.merge(df1, df2, on="DeptID",how="inner")
print(inner)
```

```python
left = pd.merge(df1, df2, on="DeptID",how="left")
print(left)
```

```python
right = pd.merge(df1, df2, on="DeptID",how="right")
print(right)
```

```python
outer = pd.merge(df1, df2, on="DeptID",how="outer")
print(outer)
```

```python
import matplotlib.pyplot as plt

data = {
    'year': [2020, 2021, 2022, 2023, 2024],
    'exchange_rate': [74.5, 73.9, 76.8, 82.5, 83.85]
}

plt.plot(data['year'], data['exchange_rate'])
plt.title("Year vs Exchange Rate")
plt.xlabel("Year")
plt.ylabel("Exchange Rate")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C++", "C"]
students = [40, 30, 20, 10]

explode = (0, 0.2, 0, 0)   # Separate the Java slice

plt.pie(students,
        labels=languages,
        autopct="%1.1f%%",
        explode=explode,
        startangle=90)

plt.title("Programming Languages")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Language": ["Python", "Java", "C++", "C"],
    "Students": [40, 30, 20, 10]
}

df = pd.DataFrame(data)

sns.barplot(x="Language", y="Students", data=df)
plt.title("Programming Languages")
plt.show()
```
